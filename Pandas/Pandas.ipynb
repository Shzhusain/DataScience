```python
# Type of Measurment

# 1. Nominal Level (Categorical)
# Definition: Data is classified into categories or names, without any order.
# Examples: Gender (Male/Female), Blood type (A, B, AB, O), Colors (Red, Blue, Green).

# 2. Ordinal Level (Ranked Order)
# Definition: Data is arranged in a meaningful order, but the differences between values are not measurable.
# Examples: Rankings (1st, 2nd, 3rd), Satisfaction levels (Satisfied, Neutral, Unsatisfied), Education level (High school, Bachelor, Master, PhD).

# 3. Interval Level
# Definition: Data is numeric, differences between values are meaningful, but there is no true zero.
# Examples: Temperature in Celsius/Fahrenheit, IQ scores, Calendar years.

# 4. Ratio Level
# Definition: Data is numeric with meaningful differences and a true zero.
# Examples: Height, Weight, Age, Income, Distance.

```

```python
# 🔹 Types of Data

# 1. Qualitative Data (Categorical / Non-numeric)

# This type of data represents qualities, labels, or categories — not numbers.

# Nominal → Categories without order.
# Example: Gender (Male/Female), Colors (Red/Blue/Green).

# Ordinal → Categories with a meaningful order, but no measurable difference.
# Example: Education level (High school < Bachelor < Master < PhD), Satisfaction (Low, Medium, High).

# 2. Quantitative Data (Numeric)

# This type of data is numerical and can be measured or counted. It has two subtypes:

# a) Discrete Data (No decimals)
# Whole numbers only, counts of items.
# Example: Number of students in a class (25, 30, 35), Number of cars in parking.

# b) Continuous Data (With decimals)
# Can take any value, including fractions/decimals.
# Example: Height (5.8 ft), Weight (70.5 kg), Temperature (36.7 °C).

# 🟡Continuous data is further divided into two measurement scales:

# Interval Scale → Numeric data with equal intervals but no true zero.
# Example: Temperature in Celsius/Fahrenheit, Dates on a calendar.

# Ratio Scale → Numeric data with equal intervals and a true zero.
# Example: Weight, Height, Age, Salary.
```

```python
# 🔹 Types of Data Analysis
# 1. Descriptive Analysis

# What it does: Summarizes past data to show what happened.
# Techniques: Averages, percentages, charts, dashboards.
# Example: “Sales increased by 20% last quarter.”

# 2. Diagnostic Analysis

# What it does: Looks deeper into the data to find out why it happened.
# Techniques: Drill-down, correlations, root cause analysis.
# Example: “Sales dropped because of supply chain delays.”

# 3. Predictive Analysis

# What it does: Uses statistical models and machine learning to predict what is likely to happen.
# Techniques: Regression, forecasting, classification, time-series analysis.
# Example: “Next quarter’s sales are predicted to rise by 15%.”

# 4. Prescriptive Analysis

# What it does: Suggests actions and strategies to influence future outcomes (tells what should be done).
# Techniques: Optimization, simulation, decision trees, AI-based recommendations.
# Example: “Increase marketing in region A to maximize profit.”
```

✅Lecture 3

Topic: Missing Values – Handling and Imputation

When working with datasets, missing values are very common.
We need to decide whether to fill (impute) them or remove the column/rows.

Rule of Thumb (Important):

If any column has less than 30% missing values, we fill (impute) the missing values.
If any column has more than 50% missing values, we remove that column from the dataset.


```python
titanic.isnull().sum() / len(titanic)*100
```

```python
#drop

# first we remove the column who has more than 50% missing values.
# meaning of axis=1 is that we are dropping a column or axis=0 is that we are dropping a row.
#inplace=true means that we are modifying the original DataFrame.
titanic.drop('deck', axis=1, inplace=True)
```

```python
titanic.isnull().sum()
```

```python
#fillna

#Now we fill the missing values of any column whos missing values less then 50% with the mean of that column.

#we have three colums in this dataset who has missung values

titanic['age'] = titanic['age'].fillna(titanic['age'].mean())
titanic['embark_town'] = titanic['embark_town'].fillna(titanic['embark_town'].mode()[0])
titanic['embarked'] = titanic['embarked'].fillna(titanic['embarked'].mode()[0])
```

```python
sns.heatmap(titanic.isnull())
```

**Feature Engineering** 

**Feature Engineering is the process of selecting and transforming raw data into features that are more suitable for modeling. The goal is to create features that are informative, relevant, and useful for the machine learning algorithm to learn from**

```python
#Binning

#Now we working on Binning 
#Main target of Binning is to divide numeric data into different groups or categories with bins and labels.
#Im going to handle just age column.
# titanic['age'].min()
# titanic['age'].max()

# we have age in the range of 0 to 80 we divided it into different categories with the help of binning,
# then we store this data into new column
bins = [0,1,5,12,18,30,50,80]
labels = ['Infant','Child','Teenager','Young Adult','Adult','Mid-Age','old']
#now we tell which column we r going to convert into bins
titanic['age_label'] = pd.cut(titanic['age'], bins=bins , labels=labels)
titanic['age_label'].value_counts()
```

```python
#rename

# If i want to rename the column 
titanic.rename(columns={'age_label':'age_labels'}, inplace=True)
titanic.head(3)
```

```python
#groupby

titanic.groupby(['age_labels','sex'])['survived'].sum()
```
