```python
# Data Scaling Techniques

# Types of Data Scaling

# 01. Standardization (Z-score Normalization) (-3 to 3)(But center will be 0 ) ✅✨(Sub sa zyada ya use hota ha)
# 02. Min-Max Scaling (scale between 0 to 1) ✅
# 03. Mazimum Absolute Scaling (scale between -1 to +1) ✅
# 04. Robust Scaling (using median and IQR)(and ignore outliers) ✅✨(agr outliers ho to use kro)
```

```python
# Importing Libraries

import pandas as pd
import numpy as np
```

```python
# Now we create a sample dataset to demonstrate data scaling techniques.


data = {
    'age_years': [15, 25, 35, 45, 55],
    'Height_cm': [150, 160, 170, 180, 190],
    'weight_kg': [50, 60, 70, 80, 90]
}   
df = pd.DataFrame(data)
df.head()
```

```python
# 01. Standardization (Z-score Normalization) ✅

# now we apply Standard Scaler to the dataset.
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaler_values = scaler.fit_transform(df)
scaler_values

# now we convert data into pandas Dataframe

# data_frame = pd.DataFrame(scaler_values, columns=df.columns)
data_frame = pd.DataFrame(scaler_values, columns=['age','Hight','Weight'])
data_frame.head()

```

```python
# 02. Min-Max Scaling (scale between 0 to 1) ✅

data = {
    'age_years': [15, 25, 35, 45, 55],
    'Height_cm': [150, 160, 170, 180, 190],
    'weight_kg': [50, 60, 70, 80, 90]
}   
df = pd.DataFrame(data)
df.head()


from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()
scaler_values = scaler.fit_transform(df)
scaler_values
data_frame = pd.DataFrame(scaler_values, columns=['age','Hight','Weight'])
data_frame.head()



```

```python
# 03. Mazimum Absolute Scaling  ✅

data = {
    'age_years': [15, 25, 35, 45, 55],
    'Height_cm': [150, 160, 170, 180, 190],
    'weight_kg': [50, 60, 70, 80, 90]
}   
df = pd.DataFrame(data)
df.head()

from sklearn.preprocessing import MaxAbsScaler
scaler = MaxAbsScaler()
scaler_values = scaler.fit_transform(df)
scaler_values
data_frame = pd.DataFrame(scaler_values, columns=['age','Hight','Weight'])
data_frame.head()



```

```python

# 04. Robust Scaling (using median and IQR) ✅

data = {
    'age_years': [15, 25, 35, 45, 55],
    'Height_cm': [150, 160, 170, 180, 190],
    'weight_kg': [50, 60, 70, 80, 90]
}   
df = pd.DataFrame(data)
df.head()

from sklearn.preprocessing import RobustScaler
scaler = RobustScaler()
scaler_values = scaler.fit_transform(df)
scaler_values
data_frame = pd.DataFrame(scaler_values, columns=['age','Hight','Weight'])
data_frame.head()
```
