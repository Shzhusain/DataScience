```python
# import Libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Data Merging

# Now we create two sample datasets to demonstrate data merging

data1 = {
    'ID': [1, 2, 3, 4], 
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'age': [25, 30, 35, 40]
}


data2 = {
    'ID': [1, 2, 3, 4],
    'city': ['New York', 'Los Angeles', 'Chicago', 'Houston'],
    'salary': [70000, 80000, 90000, 100000]
}


df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
df1.head()
```

```python
#Now we will merge these two datasets based on the 'ID' column
merged = pd.merge(df1,df2,on='ID',how='inner')
merged.head()
```

```python
# Data Concatenation
# Data Concatenation is the process of combining two or more datasets along a particular axis (rows or columns). This is useful when you have datasets with the same structure and want to combine them into a single dataset.

# Now we create three sample datasets to demonstrate data concatenation


data1 = {'Name': ['Alice', 'Bob'], 'Age': [25, 30]}
data2 = {'Name': ['Charlie', 'David'], 'Age': [35, 40]}
data3 = {'City': ['New York', 'Los Angeles'], 'Salary': [70000, 80000]}


df1 = pd.DataFrame(data1)
df2 = pd.DataFrame(data2)
df3 = pd.DataFrame(data3)
df1.head()
```

```python
# Now we will concatenate these datasets along columns

concatenated_columns = pd.concat([df1,df2],axis=1,)
concatenated_columns.head()
```

```python
# Concatenate datasets along rows
concatenated_rows = pd.concat([df1,df2],axis=0)
concatenated_rows.head()
```
