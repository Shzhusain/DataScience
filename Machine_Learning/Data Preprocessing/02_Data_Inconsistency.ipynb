```python
# Libraries 
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# we will create a sample dataframe with some inconsistencies



data = {
    'date': ['2020-1-01', '01/march/2020', '2020-03-01', '2020-04-01', '05-01-2020', '2020/06/01', '2020-07-01'],
    'Country': ['USA', 'U.S.A', 'Canada', 'Canada', 'America', 'Canada', 'USA'],
    'Name': ['John', 'john', 'jhon', 'john', 'jhon', 'john', 'john'],
    'Sales_2020': [100, 150, 200, 250, None, 350, 400],
    'Sales_2021': [110, None, 210, 260, 310, 360, 410] 
}
df = pd.DataFrame(data)
df.head() 
```

```python
# 01 Standardizing Date Formats ✅


# i want data in year-month-day format
# i use dateparser library to parse different date formats
# dateparser is more flexible than pandas to parse dates

import warnings
warnings.filterwarnings("ignore", category=DeprecationWarning)


import dateparser

def clean_date(x):
    if pd.isna(x):
        return None
    x = str(x).strip()
    result = dateparser.parse(x)
    return result

df['date'] = df['date'].apply(clean_date)
df['date'] = pd.to_datetime(df['date'], errors='coerce')
df['date'] = df['date'].dt.strftime('%Y-%m-%d')


df.head()

```

```python
# 2 inconsistent Naming Conventions(e.g., "NY" vs "New York") ✅
# we use mapping to standardize country names


country_mapping = {
'USA': 'United States','U.S.A': 'United States','America': 'United States'}
df['Country'] = df['Country'].replace(country_mapping)
df.head()
```

```python
# 3 Typographical Errors(e.g., "Calfornia" instead of "California") ✅
# we use replace method to fix typos

df['Name']= df['Name'].replace({'jhon':'john'})
df.head()
```
