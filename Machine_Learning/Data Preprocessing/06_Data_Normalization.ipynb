```python
# First we import Libraries

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# we generate some non-normal data for demonstration


np.random.seed(0)
df = np.random.exponential(size=1000,scale=2)
df = pd.DataFrame(df,columns=['value'])
df.head()

```

```python
# Now we draw the distribution of this data

sns.histplot(df['value'], kde=True)
plt.show()
```

This grapth show that the data is not normally distributed now we apply different normalization techniques to make it more normally distributed.

Different Techniques are

# 01 Box-Cox Transformation
# 02 Yeo-Johnson Transformation
# 03 Quantile Transformation

```python
# 01 Box-Cox Transformation ✨

# Data must be positive for Box-Cox Transformation

from sklearn.preprocessing import PowerTransformer

boxcox_transformer = PowerTransformer(method='box-cox')
df_boxcox = boxcox_transformer.fit_transform(df[['value']])
df_boxcox = pd.DataFrame(df_boxcox, columns=['value_boxcox'])

df_boxcox.head()
sns.histplot(df_boxcox['value_boxcox'], kde=True)
plt.show()

```

```python
# 02 Yeo-Johnson Transformation ✨
# Data can be positive or negative for Yeo-Johnson Transformation

from sklearn.preprocessing import PowerTransformer

yeojohnson_transformer = PowerTransformer(method='yeo-johnson')
df_yeojohnson = yeojohnson_transformer.fit_transform(df[['value']])
df_yeojohnson = pd.DataFrame(df_yeojohnson, columns=['value_yeojohnson'])
df_yeojohnson.head()
sns.histplot(df_yeojohnson['value_yeojohnson'], kde=True)
plt.show()
```

```python
# 03 Quantile Transformation ✨
# Data can be positive or negative for Quantile Transformation

from sklearn.preprocessing import QuantileTransformer
quantile_transformer = QuantileTransformer(output_distribution='normal')
df_quantile = quantile_transformer.fit_transform(df[['value']])
df_quantile = pd.DataFrame(df_quantile, columns=['value_quantile'])
df_quantile.head()
sns.histplot(df_quantile['value_quantile'], kde=True)
plt.show()
```

# Data Normalization 

# 01 L1 
# 02 L2



```python
# create a sample dataset
data = {'feature1': [1, 2, 3, 4, 5],
        'feature2': [10, 20, 30, 40, 50],
        'feature3': [100, 200, 300, 400, 500]}

df_norm = pd.DataFrame(data)
df_norm.head()
```

```python
from sklearn.preprocessing import Normalizer
# L1 Normalization
l1_normalizer = Normalizer(norm='l1')
df_l1_normalized = l1_normalizer.fit_transform(df_norm)
df_l1_normalized = pd.DataFrame(df_l1_normalized, columns=df_norm.columns)
df_l1_normalized.head()

```

```python
# 02 L2 Normalization ✨

l2_normalizer = Normalizer(norm='l2')
df_l2_normalized = l2_normalizer.fit_transform(df_norm)
df_l2_normalized = pd.DataFrame(df_l2_normalized, columns=df_norm.columns)
df_l2_normalized.head()

```
