```python
import numpy as np
#i wihich array print 3 rows and 2 columns with empty values
np.empty((3, 2))
```

```python
# in which array print 0 to 15 with step 2
a=np.arange(0,15,2)
print(a)
```

```python
#linspace make linear space between 0 to 10 and total values is 5    

b=np.linspace(0,10,5)
print(b)
```

```python
#if i want to get values in integer type
b=np.linspace(0,10,5 , dtype=np.int64)
print(b)
```

```python
#sorting an array
c=np.array([3,4,2,3,2,4,2,34,5,5,6,7,8,8,9])
c.sort()
print(c)
```

```python
#concatenate two arrays
a=np.array([1,2,3,4,5])
b=np.array([5,6,7,8,9])
c=np.concatenate((a,b))
print(c)
```

```python
#2d concatenate
a=np.array([[1,2],[3,4]])
b=np.array([[5,6],[7,8]])
c=np.concatenate((a,b))
print(c)    
```

```python
#if i want to check the dimension of array
print(c.ndim)   
```

```python
#3dimensional array
a=np.array([[[1,2],[3,4]],[[5,6],[7,8]],[[5,6],[7,8]]])
print(a)
```

```python
#if i want to check the shape of array
# 3 means 3 dimensions and (2,2) means 2 rows and 2 columns
print(a.shape)  
```

```python
#reshape the array
a=np.array([1,2,3,4,5,6,7,8 ])
b=a.reshape(4,2)
print(b)    
```

```python
# indexing and slicing
a=np.array([1,2,3,4,5,6,7,8 ])
print(a[0])  #print first element
print(a[1:5]) #print from index 1 to 4
print(a[:5])  #print from start to index 4
print(a[5:])  #print from index 5 to end
print(a[-1])  #print last element
print(a[-3:]) #print last 3 elements
print(a[::2]) #print step=2 elements
print(a[1::2]) #print from index 1 with step=2 elements
print(a[::-1]) #print reverse array
print(a[-2::-1]) #print from second last to start in reverse order
```
