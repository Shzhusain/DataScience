```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

```

```python
# 👉 Rule of thumb:

# Numeric–Numeric → scatter, line, heatmap

# Categorical–Numeric → bar, box, violin

# Categorical–Categorical → bar, mosaic, heatmap

# Single variable → histogram, pie, count plot
```

```python
# First we create dummy dataset
# for line plot
import plotly.express as px  

names = ['alice', 'bob', 'charlie']  
heights = [5.5, 6.0, 5.8]  

fig = px.line(x=names, y=heights, title='Heights of Students' ,height=400, width=600)  
fig.show()
```

```python
# for bar plot
# First we create dummy dataset
# for line plot
import plotly.express as px  

names = ['alice', 'bob', 'charlie']  
heights = [5.5, 6.0, 5.8]  

fig = px.bar(x=names, y=heights, title='Heights of Students',height=400, width=600)  
fig.show()
```

**For save the image**

```python
# for scatter plot
import plotly.express as px  

names = ['alice', 'bob', 'charlie']  
heights = [5.5, 6.0, 5.8]  

fig = px.scatter(x=names, y=heights, title='Heights of Students', height=400, width=600)  
fig.show()

#for save image png
# in sea born for resulation we use dpi 300 but in plotly we use scale=3 for better resulation
fig.write_image("fig1.png" ,scale=3)
#for save image in html
#fig.write_html("fig1.html")
```

**Now Making charts based on real datasets**

```python
df = sns.load_dataset('titanic')
# in sea born for grouping we use hue but in plotly we use color
df.dropna(inplace=True)
px.scatter(df , x='age' , y='fare' , color='class' ,size='fare')
```
